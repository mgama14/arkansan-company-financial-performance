```python
import pandas as pd

walmart = pd.read_csv("../data/walmart_financial_data.csv")
tyson = pd.read_csv("../data/tyson_financial_data.csv")
jbhunt = pd.read_csv("../data/jbhunt_financial_data.csv")
```

```python
walmart.info()
tyson.info()
jbhunt.info()
```

```python
walmart = walmart.rename(columns={
    "Assets": "total_assets",
    "equity": "total_equity"
})
```

```python
print(walmart.columns.tolist())
```

```python
financials = pd.concat([walmart, tyson, jbhunt], ignore_index=True)
```

```python
financials.info()
```

```python
financials.groupby("company")["fiscal_year"].agg(["min", "max", "count"])
```
