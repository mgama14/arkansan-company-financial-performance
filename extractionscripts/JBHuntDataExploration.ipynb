```python
import requests

url = "https://data.sec.gov/api/xbrl/companyfacts/CIK0000728535.json"

headers = {
    "User-Agent": "ArkansasFinancialAnalysis/1.0 contact: marylizgama@gmail.com"
}

response = requests.get(url, headers= headers)

print(response.status_code)
```

```python
data = response.json()

print(data.keys())
print(data["entityName"])
print(data["facts"].keys())

facts = data["facts"]["us-gaap"]
```

```python
revenue = facts["RevenueFromContractWithCustomerIncludingAssessedTax"]["units"]["USD"]

jbhunt_revenue = []

for year in range(2020, 2026):
    
    yearly_items = []

    for item in revenue:
        if (
            item["form"] == "10-K"
            and item["fp"] == "FY"
            and item["fy"] == year
        ):
            yearly_items.append(item)
    
    latest_item = max(yearly_items, key= lambda x: x["end"])
    
    jbhunt_revenue.append({
            "fiscal_year": year,
            "revenue": latest_item["val"]
        })

jbhunt_revenue
```

```python
operating_income = facts["OperatingIncomeLoss"]["units"]["USD"]

jbhunt_operating_income = []

for year in range(2020, 2026):

    yearly_items = []

    for item in operating_income:
        if (
            item["form"] == "10-K"
            and item["fp"] == "FY"
            and item["fy"] == year
        ):
            yearly_items.append(item)

    latest_item = max(yearly_items, key=lambda x: x["end"])

    jbhunt_operating_income.append({
        "fiscal_year": year,
        "operating_income": latest_item["val"]
    })

jbhunt_operating_income
```

```python
assets = facts["Assets"]["units"]["USD"]

jbhunt_assets = []

for year in range(2020, 2026):
    yearly_items = []

    for item in assets:
        if (
            item["form"] == "10-K"
            and item["fp"] == "FY"
            and item["fy"] == year
        ):
            yearly_items.append(item)

    latest_item = max(yearly_items, key=lambda x: x["end"])

    jbhunt_assets.append({
        "fiscal_year": year,
        "total_assets": latest_item["val"]
    })

jbhunt_assets
```

```python
equity = facts["StockholdersEquity"]["units"]["USD"]

jbhunt_equity = []

for year in range(2020, 2026):
    yearly_items = []

    for item in equity:
        if (
            item["form"] == "10-K"
            and item["fp"] == "FY"
            and item["fy"] == year
        ):
            yearly_items.append(item)

    latest_item = max(yearly_items, key=lambda x: x["end"])

    jbhunt_equity.append({
        "fiscal_year": year,
        "total_equity": latest_item["val"]
    })

jbhunt_equity
```

```python
assets = facts["Assets"]["units"]["USD"]
equity = facts["StockholdersEquity"]["units"]["USD"]

liabilities_data = []

for year in range(2020, 2026):

    yearly_assets = []
    yearly_equity = []

    for item in assets:
        if (
            item["form"] == "10-K"
            and item["fp"] == "FY"
            and item["fy"] == year
        ):
            yearly_assets.append(item)

    for item in equity:
        if (
            item["form"] == "10-K"
            and item["fp"] == "FY"
            and item["fy"] == year
        ):
            yearly_equity.append(item)

    latest_asset = max(yearly_assets, key=lambda x: x["end"])
    latest_equity = max(yearly_equity, key=lambda x: x["end"])

    liabilities = latest_asset["val"] - latest_equity["val"]

    liabilities_data.append({
        "fiscal_year": year,
        "total_liabilities": liabilities
    })

liabilities_data
```

```python
operating_cash_flow = facts["NetCashProvidedByUsedInOperatingActivities"]["units"]["USD"]

jbhunt_operating_cash_flow = []

for year in range(2020, 2026):
    yearly_items = []

    for item in operating_cash_flow:
        if (
            item["form"] == "10-K"
            and item["fp"] == "FY"
            and item["fy"] == year
        ):
            yearly_items.append(item)

    latest_item = max(yearly_items, key=lambda x: x["end"])

    jbhunt_operating_cash_flow.append({
        "fiscal_year": year,
        "operating_cash_flow": latest_item["val"]
    })

jbhunt_operating_cash_flow
```

```python
import pandas as pd

other_data = [
    jbhunt_operating_income,
    jbhunt_net_income,
    jbhunt_assets,
    jbhunt_equity,
    liabilities_data,
    jbhunt_operating_cash_flow
]
        
jbhunt_df = pd.DataFrame(jbhunt_revenue)

for data in other_data:
    temp_df = pd.DataFrame(data)
    jbhunt_df = jbhunt_df.merge(temp_df, on="fiscal_year")

jbhunt_df["company"] = "J.B. Hunt"

jbhunt_df.to_csv("jbhunt_financial_data.csv", index =False)
```
