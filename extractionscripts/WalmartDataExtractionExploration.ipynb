```python
import requests
from datetime import date

url = "https://data.sec.gov/api/xbrl/companyfacts/CIK0000104169.json"

headers = {
    "User-Agent": "ArkansasFinancialAnalysis/1.0 contact: marylizgama@gmail.com"
}

response = requests.get(url, headers=headers)

print(response.status_code)
```

```python
data = response.json()

print(data.keys())
print(data["entityName"])
print(data["facts"].keys())
```

```python
facts = data["facts"]["us-gaap"]

for fact in facts:
    if "Revenue" in fact:
        print(fact)
```

```python
##Making sure we are extracting the correct information for revenue. I want to make sure I am using the right metric here,
##as sometimes the labels are misleading, and to do so I am comparing the results to the official Walmart 2022 10-K

for concept in [
    "Revenues",
    "RevenueFromContractWithCustomerExcludingAssessedTax",
    "SalesRevenueNet"
]:
    print("\n---", concept, "---")

    observations = facts[concept]["units"]["USD"]

    for item in observations:
        if (
            item["form"] == "10-K"
            and item["fp"] == "FY"
            and 2020 <= item["fy"] <= 2025
            and item["start"][5:] == "02-01"
            and item["end"][:4] == str(item["fy"])
        ):
            print(item["fy"], item["start"], item["end"], item["val"])
```

```python
##Grabbing the Revenue numbers

revenue = facts["Revenues"]["units"]["USD"]

revenue_data = []
for item in revenue:
    if (
        item["form"] == "10-K"
        and item["fp"] == "FY"
        and 2020 <= item["fy"] <= 2025
        and item["start"][5:] == "02-01"
        and item["end"][:4] == str(item["fy"])
    ):
        revenue_data.append({
            "fiscal_year": item["fy"],
            "revenue":item["val"]
        })

print(revenue_data)
```

```python
for fact in facts:
    if "OperatingIncome" in fact:
          print(fact)
```

```python
##Grabbing Operatingincome

observations = facts["OperatingIncomeLoss"]["units"]["USD"]

operating_income_data = []

for item in observations:
    if(
        item["form"] == "10-K"
        and item["fp"] == "FY"
        and 2020 <= item["fy"] <= 2025
        and item["start"][5:] == "02-01"
        and item["end"][:4]==str(item["fy"])
    ):

        operating_income_data.append({
            "fiscal_year": item["fy"],
            "operating_income":item["val"]
        })

print(operating_income_data)
```

```python
for fact in facts:
    if "NetIncome" in fact:
        print(fact)
```

```python
##Grabbing Net Income

observations = facts["NetIncomeLoss"]["units"]["USD"]

net_income_data = []

for item in observations:
    if(
        item["form"] == "10-K"
        and item["fp"] == "FY"
        and 2020 <= item["fy"] <= 2025
        and item["start"][5:] == "02-01"
        and item["end"][:4]==str(item["fy"])
    ):

        net_income_data.append({
            "fiscal_year": item["fy"],
            "net_income":item["val"]
        })

print(net_income_data)
```

```python
for fact in facts:
    if "Assets" in fact:
        print(fact)
```

```python
##Assets does not contain a start date, so we only include an end date in the code

observations = facts["Assets"]["units"]["USD"]

assets_data = []

for item in observations:
    if (
        item["form"] == "10-K"
        and item["fp"] == "FY"
        and 2020 <= item["fy"] <= 2025
        and item["end"][:4] == str(item["fy"])
    ):
        assets_data.append({
            "fiscal_year": item["fy"],
            "Assets":item["val"]
        })

print(assets_data)
```

```python
for fact in facts:
    if "StockholdersEquity" in fact:
        print(fact)
```

```python
observations = facts["StockholdersEquity"]["units"]["USD"]

for item in observations:
    if (
        item["form"] == "10-K"
        and item["fp"] == "FY"
        and 2020 <= item["fy"] <= 2025
        and item["end"][:4] == str(item["fy"])
    ):
        print(item["fy"], item["end"], item["val"])
```

```python
for fact in facts:
    if "Equity" in fact:
        print(fact)
```

```python
##Grabbing the equity

observations = facts["StockholdersEquityIncludingPortionAttributableToNoncontrollingInterest"]["units"]["USD"]

equity_data = []

for item in observations:
    if (
        item["form"] == "10-K"
        and item["fp"] == "FY"
        and 2020 <= item["fy"] <= 2025
        and item["end"][:4] == str(item["fy"])
    ):
        equity_data.append({
            "fiscal_year": item["fy"],
            "equity":item["val"]
        })

print(equity_data)
```

```python
## There is no straightforward tagged liabilities, so we're going to manually calculate that figure here

assets = facts["Assets"]["units"]["USD"]
equity = facts["StockholdersEquityIncludingPortionAttributableToNoncontrollingInterest"]["units"]["USD"]

liabilities_data=[]

for asset_item in assets:
    if (
        asset_item["form"] == "10-K"
        and asset_item["fp"] == "FY"
        and 2020 <= asset_item["fy"] <= 2025
        and asset_item["end"][:4] == str(asset_item["fy"])
    ):
        for equity_item in equity:
            if (
                equity_item["form"] == "10-K"
                and equity_item["fp"] == "FY"
                and equity_item["fy"] == asset_item["fy"]
                and equity_item["end"] == asset_item["end"]
            ):
                liabilities = asset_item["val"] - equity_item["val"]

                liabilities_data.append({
                    "fiscal_year": asset_item["fy"],
                    "total_liabilities": liabilities
                })

print(liabilities_data)
```

```python
for fact in facts:
    if "Operating" in fact:
        print(fact)
```

```python
observations = facts["NetCashProvidedByUsedInOperatingActivities"]["units"]["USD"]

operating_activity_data=[]

for item in observations:
    if (
        item["form"] == "10-K"
        and item["fp"] == "FY"
        and 2020 <= item["fy"] <= 2025
        and item["start"][5:] == "02-01"
        and item["end"][:4] == str(item["fy"])
    ):
        operating_activity_data.append({
            "fiscal_year": item["fy"],
            "operating_cash_flow": item["val"]
        })

print(operating_activity_data)
```

```python
##Now... lets make the dataframe...

import pandas as pd

other_data = [
    operating_income_data,
    net_income_data,
    assets_data,
    equity_data,
    liabilities_data,
    operating_activity_data
]

walmart_df = pd.DataFrame(revenue_data)

for data in other_data:
    temp_df = pd.DataFrame(data)
    walmart_df = walmart_df.merge(temp_df, on="fiscal_year")
```

```python
walmart_df["company"] = "Walmart"
```

```python
walmart_df.to_csv("walmart_financial_data.csv", index=False)
```
